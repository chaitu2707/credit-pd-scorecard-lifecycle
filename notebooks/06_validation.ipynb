## 5. Scorecard recalibration

```python
sc_json = json.loads((M / "scorecard.json").read_text())
display(pd.Series({"recalibration_shift (logit)": sc_json["recalibration_shift"],
                   "C": sc_json["C"], "intercept": sc_json["intercept"]}).to_frame("value"))
```

The shift is fitted on validation so that the mean PD equals the validation bad rate.
Out-of-time is untouched. A shift that is large relative to the drift is a sign that
the model needs redevelopment, not only recalibration.
